```python
import pandas as pd
import numpy as np
import sympy as sp

df = pd.read_csv("Advertising.csv")

# 1. Matrix A (first 6 rows) and a column matrix
A = df.head(6).to_numpy()
tv = A[:, 0].reshape(-1, 1)
print("A:\n", A, "\nShape:", A.shape)
print("TV column shape:", tv.shape)
# 2. Transpose
print("Transpose:\n", A.T, "\nShape:", A.T.shape)   # (4, 6)

# 3. Inverse (A is not square, so using AᵀA)
S = A.T @ A
print("det:", np.linalg.det(S))
S_inv = np.linalg.inv(S)
print("Inverse:\n", S_inv)
print("Check S · S⁻¹ = I:", np.allclose(S @ S_inv, np.eye(4)))

# 4. Rank and RREF with SymPy
M = sp.Matrix(A.tolist())
R, pivots = M.rref()
print("Rank:", M.rank())
print("Pivot columns:", pivots)
```
